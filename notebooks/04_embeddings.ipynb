### Objective
Experiment with dense vector representations and cosine similarity calculations.
Examine how query embeddings align with relevant vs irrelevant text passages.


```python
import numpy as np

# Cosine similarity helper
def cosine_similarity(v1, v2):
    dot = np.dot(v1, v2)
    norm = np.linalg.norm(v1) * np.linalg.norm(v2)
    return dot / norm if norm > 0 else 0.0

# Demonstrate with deterministic normalized representations
v_query = np.array([0.9, 0.1, 0.05])
v_relevant = np.array([0.88, 0.12, 0.04])
v_irrelevant = np.array([0.05, 0.85, 0.6])

print(f"Similarity (Query <-> Relevant Chunk):   {cosine_similarity(v_query, v_relevant):.4f}")
print(f"Similarity (Query <-> Irrelevant Chunk): {cosine_similarity(v_query, v_irrelevant):.4f}")
```
