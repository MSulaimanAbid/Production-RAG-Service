# 03 - Chunking Strategy Comparison: Config A vs Config B
Portfolio 01 · Production RAG Service

### Objective
Compare two distinct chunking strategies to satisfy **PDF Section 3.D (Configuration Comparison)**:
- **Config A:** Recursive Character Text Splitter (512 chunk size, 50 overlap)
- **Config B:** Fixed-size Character Text Splitter (256 chunk size, 32 overlap)


```python
from pathlib import Path
import pdfplumber
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter

# Load text from attention_is_all_you_need.pdf
with pdfplumber.open("../corpus/pdfs/attention_is_all_you_need.pdf") as pdf:
    full_text = "\n\n".join(p.extract_text() or "" for p in pdf.pages[:5])

print(f"Extracted sample length: {len(full_text)} characters")
```

### 1. Apply Config A: Recursive Character Splitter (512 / 50)


```python
splitter_a = RecursiveCharacterTextSplitter(
    chunk_size=512,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""]
)
chunks_a = splitter_a.split_text(full_text)
print(f"Config A (Recursive 512/50) total chunks: {len(chunks_a)}")
print(f"Average chunk length: {sum(len(c) for c in chunks_a) / len(chunks_a):.1f} chars")
print("\nSample Chunk 0:\n", chunks_a[0][:200])
```

### 2. Apply Config B: Fixed-size Character Splitter (256 / 32)


```python
splitter_b = CharacterTextSplitter(
    separator=". ",
    chunk_size=256,
    chunk_overlap=32
)
chunks_b = splitter_b.split_text(full_text)
print(f"Config B (Fixed 256/32) total chunks: {len(chunks_b)}")
print(f"Average chunk length: {sum(len(c) for c in chunks_b) / len(chunks_b):.1f} chars")
```

### 3. Analysis & Evaluation Hypothesis
- **Config A** respects paragraph and sentence boundaries, avoiding fragmented clauses in mathematical and technical prose.
- **Config B** produces smaller, more granular chunks, which may have higher keyword density but lower semantic context.
