### Objective
Demonstrate robust text, heading hierarchy, and table extraction from Microsoft Word (`.docx`) documents using `python-docx`.


```python
from pathlib import Path
import docx

CORPUS_DOCX = Path("../corpus/docx/acme_global_remote_work_policy.docx")
print(f"Testing with: {CORPUS_DOCX.name} (Exists: {CORPUS_DOCX.exists()})")
```

### 1. Extract Headings and Paragraphs with Section Metadata


```python
doc = docx.Document(CORPUS_DOCX)
current_section = "Introduction"
sections_data = []

for para in doc.paragraphs:
    text = para.text.strip()
    if not text:
        continue
    if para.style.name.startswith("Heading"):
        current_section = text
        print(f"Found [{para.style.name}]: {text}")
    else:
        sections_data.append({"section": current_section, "text": text})

print(f"\nTotal text blocks extracted: {len(sections_data)}")
```

### 2. Extract Tables from DOCX


```python
print(f"Total tables found: {len(doc.tables)}")
for i, table in enumerate(doc.tables):
    print(f"\n--- Table {i + 1} ---")
    headers = [cell.text.strip() for cell in table.rows[0].cells]
    print(f"Headers: {headers}")
    for row in table.rows[1:3]:
        print("Row:", [c.text.strip() for c in row.cells])
```

### 3. Conclusion for Pipeline Implementation
By pairing paragraph text with the most recent `Heading` style, we can assign meaningful section names (`Section 3: Home Office Equipment Stipend`) to every extracted DOCX chunk!
